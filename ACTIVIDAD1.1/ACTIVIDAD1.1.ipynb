```python
#Importamos librerías requeridas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Carga desde un archivo .xlsx sin indice
df= pd.read_excel('Citas_Digital.xlsx')
df
```

```python
#Verificamos información del DataFrame
df.info()
```

```python
#Paso 1- Filtro por filas (0-673)

filtro1= df.iloc[0:673,: ]
filtro1.info()
```

```python
#Filtro por columnas
#Paso 2- Filtro por columnas (0,1,2,4,5,6,7,8,9)

filtro2 = filtro1.iloc[:, list(range(0, 3)) + list(range(4, 10))]
filtro2.info()
```

```python
#Paso 3- Convertir los datos de la variable “Potencial de compra” a tipo float

#Reemplazo de datos de una columna 
filtro2['Potencial de compra']=filtro2['Potencial de compra'].str.replace("Ventas ","0", regex=False)

filtro2['Potencial de compra']=filtro2['Potencial de compra'].str.replace("%","", regex=False)

filtro2['Potencial de compra']=filtro2['Potencial de compra'].str.replace("'","", regex=False)

filtro2['Potencial de compra']=filtro2['Potencial de compra'].str.replace("&","", regex=False)

#Conversión de tipo de dato de columna de tipo String a Float
filtro2['Potencial de compra']= filtro2['Potencial de compra'].astype(float)

```

```python
#Paso 4- Obtener el CSV General y nombrarlo “Citas_Digital_General.csv”  (1 entregable)

filtro2.to_csv("Citas_Digital_General.csv", index=False)
```
