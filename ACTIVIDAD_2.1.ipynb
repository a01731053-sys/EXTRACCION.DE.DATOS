```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# BDC — reemplazar con la moda

# BDC es una variable categórica, por lo que podemos reemplazar los nulos por el valor que aparece con mayor frecuencia.

```python
# Sustituir valores nulos de BDC con la moda
moda_bdc = data["BDC "].mode()[0]

data2["BDC "] = data2["BDC "].fillna(moda_bdc)
```

```python
#VERIFICAMOS SI FUNCIONO
valores_nulos=data2.isnull().sum()
valores_nulos
```

En la variable BDC se identificaron 16 valores nulos. Debido a que es una variable categórica, se decidió sustituir los valores faltantes utilizando la moda, es decir poner un valor con mayor frecuencia en la columna. En este caso, la moda fue “Ana”, por lo que los valores nulos se reemplazaron con este dato. Esta técnica  permite completar la información manteniendo la categoría más representativa de la variable.

# Potencial de compra — imputación por mediana

```python
potencial_numerico = pd.to_numeric(
    data["Potencial de compra"],
    errors="coerce"
)

potencial_numerico.median()
```

```python
mediana_potencial = potencial_numerico.median()

data2["Potencial de compra"] = data2["Potencial de compra"].fillna(
    mediana_potencial
)
```

```python
#VERIFICAMOS SI FUNCIONO
valores_nulos=data2.isnull().sum()
valores_nulos
```

Se decidió utilizar la mediana de los valores numéricos válidos, la cual fue de 0.75, ya que esta medida permite sustituir los datos faltantes sin verse tan afectada por valores atípicos o posibles errores de captura. De esta manera se conserva un valor representativo de la distribución de la variable.
